```python
import random
import numpy as np
from datasets import load_dataset, Features, Sequence, Value, ClassLabel

# Set global seed for reproducibility
SEED = 3407
random.seed(SEED)
np.random.seed(SEED)

# Original 15 NER tags used in the PET dataset
ner_tags_list = [ 
    "O", "B-Actor", "I-Actor", "B-Activity", "I-Activity", 
    "B-Activity Data", "I-Activity Data", "B-Further Specification", 
    "I-Further Specification", "B-XOR Gateway", "I-XOR Gateway", 
    "B-Condition Specification", "I-Condition Specification", 
    "B-AND Gateway", "I-AND Gateway"
]

# Define the schema to automatically cast the text tags to integers
features = Features({
    "document name": Value("string"),
    "sentence-ID": Value("int8"),
    "tokens": Sequence(Value("string")),
    "tokens-IDs": Sequence(Value("int8")),
    "ner-tags": Sequence(ClassLabel(names=ner_tags_list))
})

# Named Entity Recognition (NER) dataset
url_ner = "https://raw.githubusercontent.com/patriziobellan86/PETv1.1/master/PETv1.1-entities.jsonl"

ner_dataset = load_dataset(
    "json", 
    data_files={"test": url_ner}, 
    features=features
)

# Relation dataset
url_rel = "https://raw.githubusercontent.com/patriziobellan86/PETv1.1/master/PETv1.1-relations.json"

rel_dataset = load_dataset(
    "json", 
    data_files={"test": url_rel}
)
```

```python
# Split into 80% train, 20% test
ner_split = ner_dataset["test"].train_test_split(test_size=0.2, seed=SEED)
rel_split = rel_dataset["test"].train_test_split(test_size=0.2, seed=SEED)

print("NER dataset split:")
print(f"  Train: {len(ner_split['train'])} examples")
print(f"  Test:  {len(ner_split['test'])} examples")
print()
print("Relation dataset split:")
print(f"  Train: {len(rel_split['train'])} examples")
print(f"  Test:  {len(rel_split['test'])} examples")
```
